### 📊 Exploratory Data Analysis (EDA)

### 1️⃣ Distribution of Game Prices on Steam

```python
sns.histplot(steam['price_numeric'],bins = 20,kde = True)
plt.title('Distribution on Game Prices on Steam')
plt.xlabel('price($)')
plt.show();
```

### 📌 Insights

- Most Steam games are priced below **$20**.
- The number of games decreases as the price increases.
- High-priced games are relatively rare on the platform.
- This suggests that Steam mainly offers affordable games to attract a wider audience.

### 2️⃣ Distribution of Games Across Price Categories

```python
steam['PriceBucket'].value_counts()
```

```python
plt.figure(figsize=(10,6))
x= sns.countplot(y='PriceBucket',data=steam,palette='Blues_r')

for container in x.containers:
    x.bar_label(container)

plt.title('Distribution of Steam Games Across Price Categories')
plt.xlabel('Count')
plt.ylabel('Price_Category')

sns.despine()
plt.show();
```

### 📌 Insights

- Low-mid priced games dominate the Steam platform.
- Budget games are the second most common price category.
- Free and Premium games represent only a small portion of the dataset.
- Affordable pricing appears to be the preferred strategy for most developers.

### 3️⃣ Top 10 Developers by Number of Games Released

### 📈 Objective

Identify the top 10 developers based on the number of games released on Steam.

```python
top_dev = steam['Developers'].value_counts().head(10)
top_dev
```

```python
plt.figure(figsize=(10,6))

ax = sns.barplot(
    x=top_dev.values,
    y=top_dev.index,
    palette='viridis'
)

for container in ax.containers:
    ax.bar_label(container)

plt.title('Top 10 Developers by Number of Games Released',
          fontsize=14,
          fontweight='bold')

plt.xlabel('Number of Games')
plt.ylabel('Developers')

sns.despine()
plt.show()
```

### 📌 Insights

- CAPCOM Co., Ltd. has released the highest number of games among all developers.
- SCS Software and KOEI TECMO GAMES CO., LTD. are the next most active developers.
- The number of games drops noticeably after the top few developers.
- A small group of developers contributes a significant share of the games in the dataset.

### 4️⃣ Top 10 Publishers by Number of Games Released

```python
top_pub = steam['Publishers'].value_counts().head(10)
top_pub
```

```python
plt.figure(figsize=(10,6))

x = sns.barplot(
    x=top_pub.values,
    y=top_pub.index,
    palette='magma'
)

for container in x.containers:
    ax.bar_label(container)

plt.title('Top 10 Publishers by Number of Games Released',
          fontsize=14,
          fontweight='bold')

plt.xlabel('Number of Games')
plt.ylabel('Publishers')

sns.despine()
plt.show()
```

### 📌 Insights

- CAPCOM Co., Ltd. has published the highest number of games in the dataset.
- Paradox Interactive, Electronic Arts, and Ubisoft are also among the leading publishers.
- The top publishers have released significantly more games than the remaining publishers.
- A few major publishers contribute a large share of the games available on Steam.

### 5️⃣ Most Common Game Tags on Steam

```python
top_tags = steam['tags'].value_counts().head(10)

plt.figure(figsize=(10,6))

ax = sns.barplot(
    x=top_tags.values,
    y=top_tags.index,
    palette='crest'
)

for container in ax.containers:
    ax.bar_label(container)

plt.title(
    'Top 10 Most Common Tags on Steam',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel('Number of Games')
plt.ylabel('Tags')

sns.despine()
plt.show()
```

### 📌 Insights

- **Action** is the most common tag among Steam games.
- Many games combine multiple tags, such as **Action + Adventure** and **Action + Adventure + RPG**.
- Simulation and Strategy are also popular but appear less frequently than Action-based tags.
- This indicates that Action is the dominant game category on the Steam platform.

### 6️⃣ Relationship Between Popularity Score and Bayesian Rating

```python
plt.figure(figsize=(8,6))

sns.regplot(
    data=steam,
    x='PopularityScore',
    y='BayesianRating',
    scatter_kws={'alpha':0.4},
    line_kws={'color':'red'}
)

plt.title('Popularity Score vs Bayesian Rating',
         fontsize=14,
         fontweight='bold')

plt.xlabel('Popularity Score')
plt.ylabel('Bayesian Rating')

sns.despine()
plt.show()
```

### 📌 Insights

- There is no strong relationship between popularity score and Bayesian rating.
- Most games have ratings between **0.7** and **0.9**, regardless of their popularity.
- The regression line is nearly flat, indicating a weak correlation.
- A game can be popular without having a significantly higher rating.

### 7️⃣ Popularity of Free vs Paid Games

```python
plt.figure(figsize=(8,6))

sns.boxplot(
    data=steam,
    x='price_type',
    y='PopularityScore',
    palette='viridis'
)

plt.title(
    'Popularity of Free vs Paid Games',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel('Game Type')
plt.ylabel('Popularity Score')

sns.despine()
plt.show()
```

### 📌 Insights

- Free games have a slightly higher median popularity score than paid games.
- Paid and free games show a wide variation in popularity.
- The popularity ranges of both game types overlap considerably.
- Free games appear to have a small advantage in overall popularity.
